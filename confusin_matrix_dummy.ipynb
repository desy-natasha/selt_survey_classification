```python
import numpy as np

def confusion_matrix_from_pr(N, precision, recall, positive_ratio=0.5):
    """
    Create a binary confusion matrix from N, precision, recall, and assumed positive ratio.
    
    Parameters:
    - N: total number of samples
    - precision: precision of the classifier
    - recall: recall of the classifier
    - positive_ratio: proportion of positive samples (default 0.5)
    
    Returns:
    - cm: 2x2 confusion matrix [[TP, FP], [FN, TN]]
    """
    P = int(N * positive_ratio)          # total positives
    TP = int(round(recall * P))          # true positives
    FN = P - TP                          # false negatives
    FP = int(round(TP * (1 - precision) / precision))  # false positives
    TN = N - TP - FP - FN                # true negatives
    
    cm = np.array([[TP, FP],
                   [FN, TN]])
    return cm

# Example usage
N = 54
precision = 0.935
recall = 0.933
cm = confusion_matrix_from_pr(N, precision, recall, positive_ratio=0.4)
print(cm)
```

```python
def confusion_matrix_from_metrics(actual_positive, actual_negative, precision, recall):
    # Calculate TP
    TP = recall * actual_positive
    
    # Calculate FN
    FN = actual_positive - TP
    
    # Calculate FP
    FP = TP * (1 - precision) / precision
    
    # Calculate TN
    TN = actual_negative - FP
    
    # Optional: round to integers if needed
    TP, FP, FN, TN = map(lambda x: round(x), [TP, FP, FN, TN])
    
    return {
        'TP': TP,
        'FP': FP,
        'FN': FN,
        'TN': TN
    }

# Example usage
cm = confusion_matrix_from_metrics(actual_positive=30, actual_negative=70, precision=0.7, recall=1.00)
print(cm)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Confusion matrix values
cm = np.array([[30, 0],
               [70, 0]])

labels = ['Inappropriate', 'Appropriate']

plt.figure(figsize=(5,3))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Supervised Learning with\n Synthethic Data')
plt.yticks(rotation=0)  # keep y labels horizontal

plt.show()
```
